```python
1# Program: Pemroses Transkrip Nilai Akademik
# Penyimpanan Data Base Siswa-Siswi
if 'database_mahasiswa' not in locals():
    database_mahasiswa = []

def tampilkan_semua_data():
    """Fungsi untuk menampilkan semua mahasiswa yang tersimpan"""
    if not database_mahasiswa:
        print("\n[ Database Kosong: Belum ada data mahasiswa yang disimpan ]")
        return

    print("\n====================== SEMUA DATA MAHASISWA TERPANTAU ======================")
    print(f"| {'No':<3} | {'Nama':<15} | {'NIM':<10} | {'Tugas':<6} | {'UTS':<6} | {'UAS':<6} | {'Akhir':<6} |")
    print("----------------------------------------------------------------------------")
    for i, mhs in enumerate(database_mahasiswa, 1):
        print(f"| {i:<3} | {mhs['nama']:<15} | {mhs['nim']:<10} | {mhs['tugas']:<6.1f} | {mhs['uts']:<6.1f} | {mhs['uas']:<6.1f} | {mhs['akhir']:<6.1f} |")
    print("============================================================================")

while True:
    print("\nMENU UTAMA:")
    print("1. Input Data Mahasiswa Baru")
    print("2. Lihat Semua Data Tersimpan")
    print("3. Reset / Hapus Semua Data")
    print("4. Keluar")

    pilihan = input("Pilih menu (1/2/3/4): ").strip()

    if pilihan == '1':
        try:
            print("\n=== INPUT DATA MAHASISWA ===")
            nama = input("Masukkan Nama Mahasiswa: ").strip()
            nim = input("Masukkan NIM Mahasiswa: ").strip()
            nilai_tugas = float(input("Masukkan Nilai Tugas: "))
            nilai_uts = float(input("Masukkan Nilai UTS: "))
            nilai_uas = float(input("Masukkan Nilai UAS: "))

            # Menghitung nilai akhir
            nilai_akhir = (nilai_tugas * 0.20) + (nilai_uts * 0.35) + (nilai_uas * 0.45)

            # Menyimpan data ke database dalam bentuk dictionary
            mahasiswa_baru = {
                'nama': nama,
                'nim': nim,
                'tugas': nilai_tugas,
                'uts': nilai_uts,
                'uas': nilai_uas,
                'akhir': nilai_akhir
            }
            database_mahasiswa.append(mahasiswa_baru)
            print(f"\n[ Berhasil: Data {nama} telah disimpan! ]")

            # Menampilkan tabel ringkasan untuk data yang baru dimasukkan
            print("\n+--------------------+----------------------+------------+")
            print(f"| {'KETERANGAN':<18} | {'DATA / NILAI':<20} | {'BOBOT':<10} |")
            print("+--------------------+----------------------+------------+")
            print(f"| {'Nama':<18} | {nama:<20} | {'':<10} |")
            print(f"| {'NIM':<18} | {nim:<20} | {'':<10} |")
            print("+--------------------+----------------------+------------+")
            print(f"| {'Nilai Tugas':<18} | {nilai_tugas:<20.1f} | {0.20:<10.0%} |")
            print(f"| {'Nilai UTS':<18} | {nilai_uts:<20.1f} | {0.35:<10.0%} |")
            print(f"| {'Nilai UAS':<18} | {nilai_uas:<20.1f} | {0.45:<10.0%} |")
            print("+--------------------+----------------------+------------+")
            print(f"| {'NILAI AKHIR':<18} | {nilai_akhir:<20.1f} | {'100.0%':<10} |")
            print("+--------------------+----------------------+------------+")

        except ValueError:
            print("\nError: Input nilai (Tugas, UTS, UAS) harus berupa angka!")

    elif pilihan == '2':
        tampilkan_semua_data()

    elif pilihan == '3':
        konfirmasi = input("Apakah Anda yakin ingin menghapus SEMUA data? (y/n): ").strip().lower()
        if konfirmasi == 'y':
            database_mahasiswa.clear()
            print("\n[ Sukses: Semua data mahasiswa telah dihapus/direset! ]")
        else:
            print("\n[ Pembatalan: Data batal dihapus ]")

    elif pilihan == '4':
        print("\nProgram selesai. Terima kasih!")
        break
    else:
        print("\nPilihan tidak valid. Silakan pilih menu yang tersedia.")
```
