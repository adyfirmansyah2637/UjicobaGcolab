```python
# Program: Pemroses Transkrip Nilai Akademik
# Penyimpanan Data Base Siswa-Siswi
if 'database_mahasiswa' not in locals():
    database_mahasiswa = []

def tampilkan_semua_data():
    """Fungsi untuk menampilkan semua mahasiswa yang tersimpan"""
    if not database_mahasiswa:
        print("\n[ Database Kosong: Belum ada data mahasiswa yang disimpan ]")
        return

    print("\n====================== SEMUA DATA MAHASISWA TERPANTAU ======================")
    print(f"| {'No':<3} | {'Nama':<15} | {'NIM':<10} | {'Tugas':<6} | {'UTS':<6} | {'UAS':<6} | {'Akhir':<6} |")
    print("----------------------------------------------------------------------------")
    for i, mhs in enumerate(database_mahasiswa, 1):
        print(f"| {i:<3} | {mhs['nama']:<15} | {mhs['nim']:<10} | {mhs['tugas']:<6.1f} | {mhs['uts']:<6.1f} | {mhs['uas']:<6.1f} | {mhs['akhir']:<6.1f} |")
    print("============================================================================")

while True:
    print("\nMENU UTAMA:")
    print("1. Input Data Mahasiswa Baru")
    print("2. Lihat Semua Data Tersimpan")
    print("3. Reset / Hapus Semua Data")
    print("4. Keluar")

    pilihan = input("Pilih menu (1/2/3/4): ").strip()

    if pilihan == '1':
        try:
            print("\n=== INPUT DATA MAHASISWA ===")
            nama = input("Masukkan Nama Mahasiswa: ").strip()
            nim = input("Masukkan NIM Mahasiswa: ").strip()
            nilai_tugas = float(input("Masukkan Nilai Tugas: "))
            nilai_uts = float(input("Masukkan Nilai UTS: "))
            nilai_uas = float(input("Masukkan Nilai UAS: "))

            # Menghitung nilai akhir
            nilai_akhir = (nilai_tugas * 0.20) + (nilai_uts * 0.35) + (nilai_uas * 0.45)

            # Menyimpan data ke database dalam bentuk dictionary
            mahasiswa_baru = {
                'nama': nama,
                'nim': nim,
                'tugas': nilai_tugas,
                'uts': nilai_uts,
                'uas': nilai_uas,
                'akhir': nilai_akhir
            }
            database_mahasiswa.append(mahasiswa_baru)
            print(f"\n[ Berhasil: Data {nama} telah disimpan! ]")

            # Menampilkan tabel ringkasan untuk data yang baru dimasukkan
            print("\n+--------------------+----------------------+------------+")
            print(f"| {'KETERANGAN':<18} | {'DATA / NILAI':<20} | {'BOBOT':<10} |")
            print("+--------------------+----------------------+------------+")
            print(f"| {'Nama':<18} | {nama:<20} | {'':<10} |")
            print(f"| {'NIM':<18} | {nim:<20} | {'':<10} |")
            print("+--------------------+----------------------+------------+")
            print(f"| {'Nilai Tugas':<18} | {nilai_tugas:<20.1f} | {0.20:<10.0%} |")
            print(f"| {'Nilai UTS':<18} | {nilai_uts:<20.1f} | {0.35:<10.0%} |")
            print(f"| {'Nilai UAS':<18} | {nilai_uas:<20.1f} | {0.45:<10.0%} |")
            print("+--------------------+----------------------+------------+")
            print(f"| {'NILAI AKHIR':<18} | {nilai_akhir:<20.1f} | {'100.0%':<10} |")
            print("+--------------------+----------------------+------------+")

        except ValueError:
            print("\nError: Input nilai (Tugas, UTS, UAS) harus berupa angka!")

    elif pilihan == '2':
        tampilkan_semua_data()

    elif pilihan == '3':
        konfirmasi = input("Apakah Anda yakin ingin menghapus SEMUA data? (y/n): ").strip().lower()
        if konfirmasi == 'y':
            database_mahasiswa.clear()
            print("\n[ Sukses: Semua data mahasiswa telah dihapus/direset! ]")
        else:
            print("\n[ Pembatalan: Data batal dihapus ]")

    elif pilihan == '4':
        print("\nProgram selesai. Terima kasih!")
        break
    else:
        print("\nPilihan tidak valid. Silakan pilih menu yang tersedia.")
```

```python
# Program Utama: Sistem Database Akademik Multi-Kelas
# Menyimpan database dengan struktur: {'Nama_Kelas': [daftar_siswa]}
if 'database_inti_kelas' not in locals():
    database_inti_kelas = {}

def tampilkan_semua_kelas():
    """Menampilkan daftar semua kelas yang terdaftar"""
    if not database_inti_kelas:
        print("\n[ Sistem: Database Kosong - Belum ada kelas yang terdaftar ]")
        return []

    daftar_kls = list(database_inti_kelas.keys())
    print("\n================ DAFTAR KELAS TERDAFTAR ================")
    for idx, nama_kelas in enumerate(daftar_kls, 1):
        jumlah_siswa = len(database_inti_kelas[nama_kelas])
        print(f"{idx}. Kelas: {nama_kelas:<15} | Jumlah Siswa: {jumlah_siswa} orang")
    print("========================================================")
    return daftar_kls

def tampilkan_tabel_dan_rata_rata(nama_kelas):
    """Menampilkan tabel nilai siswa dan menghitung rata-rata nilai kelas"""
    siswa_list = database_inti_kelas.get(nama_kelas, [])
    if not siswa_list:
        print(f"\n[ Kelas {nama_kelas} masih kosong atau tidak memiliki data siswa ]")
        return

    print(f"\n====================== LAPORAN AKADEMIK KELAS: {nama_kelas.upper()} ======================")
    print(f"| {'No':<3} | {'Nama':<15} | {'NIM':<12} | {'Tugas':<6} | {'UTS':<6} | {'UAS':<6} | {'Akhir':<6} |")
    print("---------------------------------------------------------------------------------------")

    total_akhir = 0
    nilai_tertinggi = -1
    nilai_terendah = 101

    for i, mhs in enumerate(siswa_list, 1):
        akhir = mhs['akhir']
        total_akhir += akhir
        if akhir > nilai_tertinggi: nilai_tertinggi = akhir
        if akhir < nilai_terendah: nilai_terendah = akhir

        print(f"| {i:<3} | {mhs['nama']:<15} | {mhs['nim']:<12} | {mhs['tugas']:<6.1f} | {mhs['uts']:<6.1f} | {mhs['uas']:<6.1f} | {akhir:<6.1f} |")

    rata_rata = total_akhir / len(siswa_list)

    print("=======================================================================================")
    print(f"| RATA-RATA NILAI AKHIR KELAS : {rata_rata:<54.2f} |")
    print(f"| NILAI AKHIR TERTINGGI       : {nilai_tertinggi:<54.2f} |")
    print(f"| NILAI AKHIR TERENDAH        : {nilai_terendah:<54.2f} |")
    print("=======================================================================================")

def dapatkan_pilihan_kelas(input_user, daftar_kls):
    """Membantu mencocokkan input user baik berupa nomor indeks maupun nama kelas"""
    if not input_user:
        return None

    # Jika user menginput angka indeks pilihan
    if input_user.isdigit():
        idx = int(input_user) - 1
        if 0 <= idx < len(daftar_kls):
            return daftar_kls[idx]

    # Jika user menginput nama kelas langsung
    input_upper = input_user.upper()
    if input_upper in daftar_kls:
        return input_upper

    return None

while True:
    print("\n======== DATABASE INTI AKADEMIK MULTI-KELAS ========")
    print("1. Tambah Kelas Baru")
    print("2. Input Data Siswa Baru ke Kelas")
    print("3. Lihat Laporan Nilai & Rata-rata per Kelas")
    print("4. Hapus / Kurangi Kelas")
    print("5. Reset Data di dalam Kelas (Kosongkan Kelas)")
    print("6. Reset Total Seluruh Database")
    print("7. Keluar")
    print("====================================================")

    pilihan = input("Pilih menu (1-7): ").strip()

    if pilihan == '1':
        print("\n=== TAMBAH KELAS BARU ===")
        nama_kelas_baru = input("Masukkan Nama Kelas Baru (contoh: 1A, TI-3B): ").strip().upper()
        if not nama_kelas_baru:
            print("\nError: Nama kelas tidak boleh kosong!")
        elif nama_kelas_baru in database_inti_kelas:
            print(f"\nError: Kelas '{nama_kelas_baru}' sudah terdaftar di database.")
        else:
            database_inti_kelas[nama_kelas_baru] = []
            print(f"\n[ Berhasil: Kelas '{nama_kelas_baru}' telah berhasil dibuat! ]")

    elif pilihan == '2':
        print("\n=== INPUT DATA SISWA BARU ===")
        if not database_inti_kelas:
            buat_langsung = input("Belum ada kelas terdaftar. Buat kelas baru sekarang? (y/n): ").strip().lower()
            if buat_langsung == 'y':
                nama_kelas_baru = input("Masukkan Nama Kelas Baru: ").strip().upper()
                if nama_kelas_baru:
                    database_inti_kelas[nama_kelas_baru] = []
                    target_kelas = nama_kelas_baru
                else:
                    print("Batal membuat kelas.")
                    continue
            else:
                continue
        else:
            daftar_kls = tampilkan_semua_kelas()
            input_kelas = input("\nMasukkan Nama Kelas atau Nomor Kelas Tujuan: ").strip()
            target_kelas = dapatkan_pilihan_kelas(input_kelas, daftar_kls)

        if target_kelas and target_kelas in database_inti_kelas:
            try:
                nama = input("Masukkan Nama Siswa: ").strip()
                nim = input("Masukkan NIM Siswa: ").strip()
                nilai_tugas = float(input("Masukkan Nilai Tugas: "))
                nilai_uts = float(input("Masukkan Nilai UTS: "))
                nilai_uas = float(input("Masukkan Nilai UAS: "))

                # Rumus Bobot Nilai Akhir
                nilai_akhir = (nilai_tugas * 0.20) + (nilai_uts * 0.35) + (nilai_uas * 0.45)

                siswa_baru = {
                    'nama': nama,
                    'nim': nim,
                    'tugas': nilai_tugas,
                    'uts': nilai_uts,
                    'uas': nilai_uas,
                    'akhir': nilai_akhir
                }
                database_inti_kelas[target_kelas].append(siswa_baru)
                print(f"\n[ Berhasil: {nama} resmi masuk ke Kelas '{target_kelas}' ]")
            except ValueError:
                print("\nError: Nilai akademis harus diisi dengan format angka!")
        else:
            print(f"\nError: Kelas tujuan tidak valid atau tidak ditemukan.")

    elif pilihan == '3':
        if not database_inti_kelas:
            print("\n[ Database Kosong: Daftarkan kelas terlebih dahulu ]")
            continue

        print("\n=== LAPORAN KELAS & RATA-RATA NILAI ===")
        daftar_kls = tampilkan_semua_kelas()
        input_kelas = input("\nMasukkan Nama Kelas atau Nomor Kelas yang ingin dianalisis: ").strip()
        pilih_kls = dapatkan_pilihan_kelas(input_kelas, daftar_kls)

        if pilih_kls and pilih_kls in database_inti_kelas:
            tampilkan_tabel_dan_rata_rata(pilih_kls)
        else:
            print(f"\nError: Kelas tidak ditemukan!")

    elif pilihan == '4':
        if not database_inti_kelas:
            print("\n[ Database Kosong: Tidak ada kelas untuk dihapus ]")
            continue

        print("\n=== HAPUS / KURANGI KELAS ===")
        daftar_kls = tampilkan_semua_kelas()
        input_kelas = input("\nMasukkan Nama Kelas atau Nomor Kelas yang ingin dihapus total: ").strip()
        kelas_hapus = dapatkan_pilihan_kelas(input_kelas, daftar_kls)

        if kelas_hapus and kelas_hapus in database_inti_kelas:
            konfirmasi = input(f"Apakah Anda yakin ingin menghapus kelas '{kelas_hapus}' dan semua isinya? (y/n): ").strip().lower()
            if konfirmasi == 'y':
                del database_inti_kelas[kelas_hapus]
                print(f"\n[ Sukses: Kelas '{kelas_hapus}' telah dihapus dari database ]")
        else:
            print(f"\nError: Kelas tidak ditemukan.")

    elif pilihan == '5':
        if not database_inti_kelas:
            print("\n[ Database Kosong: Tidak ada kelas untuk dikosongkan ]")
            continue

        print("\n=== KOSONGKAN ISI DATA KELAS ===")
        daftar_kls = tampilkan_semua_kelas()
        input_kelas = input("\nMasukkan Nama Kelas atau Nomor Kelas yang datanya ingin di-reset: ").strip()
        kelas_reset = dapatkan_pilihan_kelas(input_kelas, daftar_kls)

        if kelas_reset and kelas_reset in database_inti_kelas:
            konfirmasi = input(f"Apakah Anda yakin ingin menghapus semua data siswa di kelas '{kelas_reset}'? (y/n): ").strip().lower()
            if konfirmasi == 'y':
                database_inti_kelas[kelas_reset].clear()
                print(f"\n[ Sukses: Data siswa di kelas '{kelas_reset}' telah dibersihkan ]")
        else:
            print(f"\nError: Kelas tidak ditemukan.")

    elif pilihan == '6':
        if not database_inti_kelas:
            print("\n[ Sistem: Database memang sudah kosong ]")
            continue

        konfirmasi = input("\nAPAKAH ANDA YAKIN INGIN MENGHAPUS SELURUH KELAS DAN DATA? (y/n): ").strip().lower()
        if konfirmasi == 'y':
            database_inti_kelas.clear()
            print("\n[ Sukses: Database berhasil di-reset total ke kondisi awal ]")

    elif pilihan == '7':
        print("\nSistem dimatikan. Terima kasih telah menggunakan Database Akademik Inti!")
        break
    else:
        print("\nPilihan menu salah! Masukkan angka antara 1 sampai 7.")
```
